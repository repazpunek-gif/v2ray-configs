```python
from diffusers import StableDiffusionPipeline
import torch

# Загрузка модели Stable Diffusion. Это займет некоторое время при первом запуске.
# Для использования этой модели может потребоваться авторизация на Hugging Face.
# Убедитесь, что вы приняли условия использования модели на странице Hugging Face:
# https://huggingface.co/runwayml/stable-diffusion-v1-5
# и вошли в систему, используя 'huggingface-cli login' в терминале Colab

model_id = "runwayml/stable-diffusion-v1-5"
pipeline = StableDiffusionPipeline.from_pretrained(model_id, torch_dtype=torch.float16)
pipeline.to("cuda")

# Пример промта для генерации изображения
prompt = "a photo of an astronaut riding a horse on mars"
image = pipeline(prompt).images[0]

# Сохранение сгенерированного изображения
image.save("astronaut_horse.png")
print("Сгенерированное изображение сохранено как astronaut_horse.png")
```
